# CSV to Parquet Data Conversion

```python
import os
import pandas as pd

project_root = os.path.dirname(os.getcwd())
csv_path = os.path.join(project_root, "data", "raw", "financial_statement.csv")
parquet_path = os.path.join(project_root, "data", "processed", "financial_statement.parquet")

data = pd.read_csv(csv_path, encoding="utf-8-sig")
data.columns = data.columns.str.strip()
data["Category"] = data["Category"].str.upper()
data.to_parquet(parquet_path, index=False)

print(f"Converted {len(data)} rows, {len(data.columns)} columns")
print(f"Saved to: {parquet_path}")
```
